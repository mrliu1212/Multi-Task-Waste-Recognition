```python
import os
os.environ["WANDB_DISABLED"] = "true"
from ultralytics import YOLO

# Path to dataset
yaml_directory = "taco_yolo_4cats/taco.yaml"

model = YOLO('yolov8s.pt') 

# Train the model
results = model.train(
    data=yaml_directory,
    
    # GENERAL SETTINGS
    epochs=500,               
    patience=30,              
    batch=8,                 
    name="yolov8s_taco_opt",
    project="taco_yolo_4cats_runs",
    seed=42,
    
    # IMAGE SETTINGS
    imgsz=1024,               # HIGHEST RESOLUTION THAT FITS IN GPU MEMORY.
    
    # AUGMENTATION
    # Trash has no specific orientation (unlike a car or a person).
    degrees=45.0,             # Rotate images +/- 45 degrees.
    flipud=0.5,               # Vertical Flip: 50% chance. Trash can be upside down.
    fliplr=0.5,               # Horizontal Flip: 50% chance (Standard).
    mosaic=1.0,               # IT PUTS 4 IMAGES TOGETHER TO MAKE A MOSAIC.
    
    # OPTIMIZATION
    optimizer='auto',        
    cos_lr=True,              # ENSURE SMOOTH LR DECAY.
    close_mosaic=10,          # FOR LAST 10 EPOCHS, NO MOSAIC AUGMENTATION.
)
```
