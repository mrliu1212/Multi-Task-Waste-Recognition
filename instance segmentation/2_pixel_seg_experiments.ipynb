### Pixel Segmentation (Experiments)

```python
from pathlib import Path
import yaml
from collections import Counter
import matplotlib.pyplot as plt
import numpy as np

PROJECT_ROOT = Path("/home/default/Desktop/project")
DATASET_DIR = PROJECT_ROOT / "Dataset 4cats" / "dataset_yolo_4cats_seg"
DATA_YAML = DATASET_DIR / "taco.yaml"

EXPERIMENT_DIR = PROJECT_ROOT / "taco_seg_experiments"
EXPERIMENT_DIR.mkdir(exist_ok=True)

print(f"Dataset dir: {DATASET_DIR} -> {DATASET_DIR.exists()}")
print(f"Data YAML: {DATA_YAML} -> {DATA_YAML.exists()}")
```

#### Experiment 1: Baseline with Cosine LR

```python
from ultralytics import YOLO
model = YOLO("yolov8s-seg.pt")

results_exp1 = model.train(
    data=str(DATA_YAML),
    epochs=150,
    imgsz=640,
    batch=8,
    project=str(EXPERIMENT_DIR),
    name="exp1_baseline_cosine",
    cos_lr=True,          
    lr0=0.01,
    lrf=0.001,            
    warmup_epochs=5,
    patience=30,          
    pretrained=True,
    verbose=True,
    deterministic=True,
    seed=42,
)
```

#### Experiment 2: Medium Model (yolov8m-seg)

```python
model = YOLO("yolov8m-seg.pt")

results_exp2 = model.train(
    data=str(DATA_YAML),
    epochs=150,
    imgsz=640,
    batch=6,              
    project=str(EXPERIMENT_DIR),
    name="exp2_medium_model",
    cos_lr=True,
    lr0=0.01,
    lrf=0.001,
    warmup_epochs=5,
    patience=30,
    pretrained=True,
    verbose=True,
)
```

#### Experiment 3: Higher Resolution (768px)

```python
model = YOLO("yolov8m-seg.pt")

results_exp3 = model.train(
    data=str(DATA_YAML),
    epochs=150,
    imgsz=768,
    batch=4,
    project=str(EXPERIMENT_DIR),
    name="exp3_higher_resolution",
    cos_lr=True,
    lr0=0.01,
    lrf=0.001,
    warmup_epochs=5,
    patience=30,
    pretrained=True,
    verbose=True,
)
```

#### Experiment 4: Copy-Paste Augmentation

```python
model = YOLO("yolov8m-seg.pt")

results_exp4 = model.train(
    data=str(DATA_YAML),
    epochs=100,
    imgsz=640,
    batch=8,
    project=str(EXPERIMENT_DIR),
    name="exp4_copy_paste",
    copy_paste=0.5,
    copy_paste_mode="flip",
    
    # Additional augmentation
    mixup=0.1,
    mosaic=1.0,
    close_mosaic=15,
    
    # Learning rate
    cos_lr=True,
    lr0=0.01,
    lrf=0.001,
    warmup_epochs=5,
    patience=30,
    
    pretrained=True,
    verbose=True,
)
```
