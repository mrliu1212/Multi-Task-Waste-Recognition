```python
from pathlib import Path 

data_root = Path("TACO/data")  # or Path("/content/TACO/data") if you're unsure of cwd
image_exts = {".jpg", ".jpeg", ".png"}  # add more if needed

def count_images(folder: Path) -> int:
    return sum(1 for p in folder.rglob("*") if p.suffix.lower() in image_exts)

total = 0
for batch in sorted(data_root.glob("batch_*")):
    n = count_images(batch)
    total += n
    print(f"{batch.name}: {n} images")
print(f"TOTAL: {total} images")
```
