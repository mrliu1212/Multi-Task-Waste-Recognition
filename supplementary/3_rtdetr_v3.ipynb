#### Train RT-DETR

```python
import os
os.environ["WANDB_DISABLED"] = "true"
from ultralytics import RTDETR

model = RTDETR('rtdetr-l.pt')

results = model.train(
    data="dataset_yolo_4cats_seg/taco_oversampled.yaml",
    
    epochs=100,
    batch=8,
    imgsz=1024,

    project="Yolo_training",
    name="rtdetr_1024_oversampled",
    
    optimizer='AdamW',
    lr0=0.0001,
    
    # Augmentation (Trash specific tweaks)
    degrees=45.0,
    flipud=0.5,
    fliplr=0.5,
    mosaic=0.2,
    mixup=0.25,
    copy_paste=0.5, 
    hsv_h=0.015,
    hsv_s=0.7,
    hsv_v=0.4,
)
```
